# Real-World Example: Fraud Detection

Let's look at how a bank tries to catch fraudulent credit card transactions using both approaches.

---

### Version 1: The Rule-Based Approach (Traditional Programming)

A software engineer sits down with a banking expert and manually writes a set of rules based on common fraud patterns.

```python
# Manually coded rules
def check_transaction(amount, location, time_of_day, typical_location):
    # Rule 1: Flag if the transaction amount is unusually high
    if amount > 10000:
        return "Flag as Fraud"
    
    # Rule 2: Flag if the transaction happens in a different country within an impossible timeframe
    if location != typical_location and time_of_day == "3:00 AM":
        return "Flag as Fraud"
    
    # Otherwise, let it pass
    return "Approved"

- What if a legitimate user goes on vacation and buys an expensive dinner at 3:00 AM? The rule blocks them.
- What if a clever fraudster steals $500 fifty times in a row from the user's home city? The rules completely miss it.
- As fraudsters change their tactics, humans must constantly write, test, and patch hundreds of new rules manually. It quickly becomes an impossible nightmare.

Past Transaction Data (Amounts, Locations, Times, Device IDs) 

+ 

Correct Labels (Fraud / Legitimate)

        ↓

Learning Algorithm (Discovers subtle, multi-layered combinations)

        ↓

Trained Model (Can flag complex fraud patterns humans wouldn't notice)

#### Why Version 2 Wins:
- The computer automatically realizes that fraud isn't just about a single high amount or a weird time;
- It might be a subtle combination of Device Type + Rapid Successive Micro-transactions + Specific Category of Merchant.
- If fraudsters shift their behavior, we don't rewrite code. We simply retrain the model with the latest data, and it updates its own internal parameters automatically.

# The Idea Behind Machine Learning

Instead of writing the rules ourselves...

We provide:

- Many examples
- Input data
- Correct answers

The computer automatically discovers the underlying patterns.

```

IF temperature > 38°C
AND cough = Yes
THEN Possible Flu
```

These systems do not learn from experience.

Instead, experts manually define all decision rules.

This approach is known as a **Rule-Based System** or **Expert System**.

---

# What is Machine Learning (ML)?

Machine Learning is a subset of Artificial Intelligence.

Instead of writing explicit rules, we provide:

- Data
- Examples
- Expected outputs (sometimes)

The computer automatically discovers patterns in the data.

> **Machine Learning is the science of learning patterns from data.**

Instead of programming every rule, we allow the algorithm to learn them.

---

# Traditional Programming vs Machine Learning

### Traditional Programming

```
Rules + Data
        ↓
    Program
        ↓
    Answers
```

Humans write the rules.

---

### Machine Learning

```
Data + Answers
        ↓
Learning Algorithm
        ↓
      Model
```

The computer learns the rules automatically.

---

# Examples of Machine Learning

Machine Learning is widely used in:

- Spam detection
- House price prediction
- Credit scoring
- Product recommendation
- Fraud detection
- Customer segmentation

These systems improve as they receive more data.

---

# What is Deep Learning (DL)?

Deep Learning is a specialized branch of Machine Learning.

Instead of using simple mathematical models, it uses **Artificial Neural Networks** with many layers.

These networks can automatically learn complex representations from large amounts of data.

> **Deep Learning uses multi-layer neural networks to learn complex patterns from data.**

Deep Learning has become the dominant approach for many AI applications.

---

# Why Deep Learning?

Deep Learning performs exceptionally well on complex data such as:
- Images
- Speech
- Video
- Natural Language
- Large-scale datasets

It has enabled breakthroughs in:

- Face recognition
- Speech recognition
- Machine translation
- Autonomous driving
- Large Language Models (LLMs)


# The Scale and Drivers of Deep Learning

Deep Learning performs exceptionally well on complex, unstructured data because of two main drivers:
1. **The Data Explosion:** Traditional ML algorithms plateau in performance, but DL models keep improving as you feed them more data.
2. **Hardware (GPUs):** Massive computational power allows us to train billions of parameters.

---

# Relationship Between AI, ML, and DL

The relationship can be visualized as nested circles:

```

```

Category:

**Artificial Intelligence**

(Not Machine Learning)

---

## Example 2: Spam Filter

The system learns from thousands of labeled emails.

It improves automatically as new examples become available.

Category:

**Machine Learning**

---

## Example 3: Face Recognition

The system analyzes millions of images and learns facial representations automatically.

Category:

**Deep Learning**

---

## Example 4: ChatGPT

ChatGPT is built using very large deep neural networks called **Transformer models**.

It learns language patterns from enormous amounts of text.

Category:

**Deep Learning**

(It is also a Machine Learning and Artificial Intelligence system.)

---

# Practice

Classify each application into the correct category.

| Application | AI | ML | DL |
|--------------|:--:|:--:|:--:|
| Chess expert system | ✓ | | |
| Spam filter | ✓ | ✓ | |
| House price prediction | ✓ | ✓ | |
| Face recognition | ✓ | ✓ | ✓ |
| ChatGPT | ✓ | ✓ | ✓ |
| Rule-based medical diagnosis | ✓ | | |

---

# Key Takeaways

- Artificial Intelligence is the broadest field.
- Machine Learning is one approach for building AI systems.
- Deep Learning is a specialized type of Machine Learning.
- Not every AI system uses Machine Learning.
- Every Deep Learning model is also a Machine Learning model and an Artificial Intelligence system.

In the next section, we will learn about the most important ingredient of every Machine Learning system:

**Data.**

# Data, Samples, Features, and Labels

## Learning Objectives

By the end of this section, you should be able to:

- Understand the role of data in Machine Learning
- Define Dataset, Sample, Feature, and Label
- Identify features and labels in real-world problems
- Understand the difference between Classification and Regression problems

---

## Why is Data Important in Machine Learning?

Data is the foundation of every Machine Learning system.

A Machine Learning model learns patterns from data and uses those patterns to make predictions or decisions.

A simple way to think about Machine Learning is:

Data

↓

Learning Process

↓

Model

↓

Prediction



Without data, a Machine Learning model has nothing to learn from.

---

# What is a Dataset?

A **Dataset** is a collection of data examples used for training, evaluating, or testing a Machine Learning model.

A dataset usually contains:

- Multiple examples (samples)
- Information about each example (features)
- Sometimes the expected answer (label)

Example:

A dataset of house prices:

| Area (m²) | Bedrooms | Age | Price |
|-----------|----------|-----|-------|
| 120 | 3 | 10 | 350000 |
| 90 | 2 | 20 | 220000 |
| 150 | 4 | 5 | 500000 |

This entire table is a **Dataset**.

---

# What is a Sample?

A **Sample** is a single example or observation inside a dataset.

Each row in a dataset represents one sample.

Example:

From the house price dataset:

| Area | Bedrooms | Age | Price |
|------|----------|-----|-------|
| 120 | 3 | 10 | 350000 |

This row represents **one sample**.

In Machine Learning terminology:


Dataset = Collection of Samples

Sample 1

Sample 2

Sample 3

...

Sample N

---

# What is a Feature?

A **Feature** is an input variable that describes a sample.

Features generally fall into two main types:
1. **Numerical Features:** Continuous or discrete numbers (e.g., Area: 120m², Bedrooms: 3).
2. **Categorical Features:** Labels or categories representing distinct groups (e.g., Location: "City Center", "Suburb").



Features are the information that the model uses to make predictions.

Features are usually represented as numbers, categories, or other measurable information.

Example: Predicting house prices

| Feature | Description |
|---------|-------------|
| Area | Size of the house |
| Bedrooms | Number of rooms |
| Age | Years since construction |

The model uses these features to predict the house price.

---

# What is a Label?

A **Label** is the output or target value that we want the model to predict.

The label is also called:

- Target
- Output
- Ground Truth

Example: Predicting house prices

| Area | Bedrooms | Age | Price |
|------|----------|-----|-------|
| 120 | 3 | 10 | 350000 |

Features:
Area, Bedrooms, Age

Label(s): Price


The goal of the model is to learn the relationship between features and labels.

---

# Features and Labels: General Structure

A typical Machine Learning dataset looks like this:


        Features              Label

    Input Variables            Expected Output

          X                        y

[x1, x2, x3]   ----->      y


### During training, the model learns:

Features → Label

#### During the training step, an optimal map/function between features and label is obtained.

### Then, after learning:

New Features → Predicted Label

#### During the inference step, the achieved map/function is used to predict label for every given feature.


---

# Example 1: House Price Prediction

## Problem

Predict the price of a house based on its characteristics.

Dataset:

| Area | Bedrooms | Location | Price |
|------|----------|----------|-------|
| 120 | 3 | City Center | 350000 |
| 80 | 2 | Suburb | 180000 |

Features:

- Area
- Number of bedrooms
- Location

Label:

- House price

Question:

Input:

Area = 100 m²
Bedrooms = 2
Location = Suburb

Output:

Predicted Price


---

# Example 2: Disease Detection

## Problem

Predict whether a patient has a specific disease.

Dataset:

| Age | Blood Pressure | Symptoms | Disease Status |
|-----|----------------|----------|---------|
| 45 | High | Yes | Positive |
| 30 | Normal | No | Negative |

Features:

- Age
- Blood pressure
- Symptoms

Label:

- Disease status

The model learns:

Patient Information → Disease Prediction



---

# Example 3: Student Score Prediction

## Problem

Predict a student's final score.

Dataset:

| Study Hours | Attendance | Previous Score | Final Score |
|-------------|------------|----------------|-------------|
| 5 | 90% | 75 | 82 |
| 2 | 60% | 50 | 45 |

Features:

- Study hours
- Attendance
- Previous score

Label:

- Final score

The model learns:

Student Information → Expected Score



---

# Example 4: Email Classification

## Problem

Determine whether an email is spam or not.

Dataset:

| Email Text | Spam |
|------------|------|
| "Win a free prize!" | Yes |
| "Meeting tomorrow" | No |

Features:

- Words in the email
- Email characteristics

Label:

- Spam / Not Spam

The model learns:

Email Content → Spam Classification


---

# Types of Machine Learning Problems

After defining features and labels, we can categorize Machine Learning problems based on the type of output we want to predict.

The two most common categories are:

1. Classification
2. Regression

---

# Classification

Classification is a problem where the output belongs to a specific category or class.

The label is usually a discrete value.

Examples:

| Problem | Input | Output |
|---------|-------|--------|
| Spam detection | Email | Spam / Not Spam |
| Disease detection | Patient data | Positive / Negative |
| Image recognition | Image | Cat / Dog |

Output examples:

Yes / No

Spam / Not Spam

Class A / Class B / Class C


---

# Regression

Regression is a problem where the output is a continuous numerical value.

Examples:

| Problem | Input | Output |
|---------|-------|--------|
| House price prediction | House information | Price |
| Temperature prediction | Weather data | Temperature |
| Score prediction | Student information | Score |

Output examples:

350000

23.5°C

87.5



---


# Classification vs Regression

| Classification | Regression |
|----------------|------------|
| Predicts categories | Predicts numbers |
| Output is discrete | Output is continuous |
| Example: Spam or Not Spam | Example: House Price |
| Example: Disease Yes/No | Example: Temperature |

---

# Practice Exercise

For each problem, identify:

1. Features
2. Label
3. Problem type (Classification or Regression)

---

## Problem 1

Predict whether a customer will buy a product.

Possible features:

- Age
- Previous purchases
- Website visits

Label:

?

Type:

?

---

## Problem 2

Predict the price of a used car.

Possible features:

- Brand
- Year
- Mileage

Label:

?

Type:

?

---

# Key Takeaways

- Data is the foundation of Machine Learning.
- A dataset contains many samples.
- A sample represents one example.
- Features are the inputs used by the model.
- Labels are the outputs we want to predict.
- Classification predicts categories.
- Regression predicts numerical values.

In the next section, we will learn about different types of Machine Learning approaches and how models learn from data.

# Types of Machine Learning

## Learning Objectives

By the end of this section, you should be able to:

- Understand the main types of Machine Learning
- Explain the difference between Supervised, Unsupervised, and Reinforcement Learning
- Identify the appropriate learning approach for different real-world problems

---

# How Do Machines Learn?

Machine Learning systems learn from experience.

The type of experience and the available information determine the learning approach.

The three main categories of Machine Learning are:

1. Supervised Learning
2. Unsupervised Learning
3. Reinforcement Learning



---

# 1. Supervised Learning

## Definition

Supervised Learning is a type of Machine Learning where the model learns from **labeled data**.

The training dataset contains:

- Input features
- Correct output (label)

The model learns the relationship between inputs and outputs.

The goal is to predict the correct output for new, unseen data.

---

## How Supervised Learning Works

The learning process looks like this:




Features + Labels --> Learning Algorithm --> Trained Model --> Prediction for New Data 

Example:


Input: (House area = 120 m², Bedrooms = 3) --> Model --> Output: Predicted Price = 350000


---

# Types of Supervised Learning

Supervised Learning is usually divided into two main categories:

## 1. Classification

The model predicts a category.

Examples:

Email → Spam / Not Spam

Image → Cat / Dog

Patient → Disease / No Disease



---


## 2. Regression

The model predicts a numerical value.

Examples:

House → Price

Weather → Temperature

Student Information → Final Score


---


# Real-World Applications of Supervised Learning

## Spam Detection

Input:

- Email text
- Sender information
- Email features

Output:

- Spam
- Not Spam

---

## Medical Diagnosis

Input:

- Patient information
- Medical measurements

Output:

- Disease category

---

## Price Prediction

Input:

- House features
- Market information

Output:

- Estimated price

---

# 2. Unsupervised Learning

## Definition

Unsupervised Learning is a type of Machine Learning where the model learns from **unlabeled data**.

The dataset contains only input features.

There are no predefined answers.

The goal is to discover hidden patterns, structures, or relationships in the data.

---

## How Unsupervised Learning Works

Data (No Labels) --> Learning Algorithm --> Discover Patterns



The model tries to answer questions such as:

- Are there groups in this data?
- Which examples are similar?
- Are there unusual patterns?

---

# Example: Customer Segmentation

A company has information about customers:

| Age | Income | Purchase History |
|-----|--------|------------------|
| 25 | Low | Few purchases |
| 45 | High | Many purchases |
| 30 | Medium | Regular purchases |

There are no predefined customer groups.

The algorithm may discover groups such as:

Group 1:
Young customers with low spending

Group 2:
High-income frequent buyers

Group 3:
Occasional customers

---

# Common Tasks in Unsupervised Learning

## Clustering

Grouping similar data points together.

Examples:

- Customer segmentation
- Grouping news articles
- Organizing documents

---

## Dimensionality Reduction

Reducing the number of features while preserving important information.

Examples:

- Data visualization
- Feature compression

---

# Real-World Applications of Unsupervised Learning

## Customer Segmentation

Companies group customers based on behavior.

Used for:

- Marketing strategies
- Personalized recommendations

---

## Recommendation Systems

Finding similar users or products.

Examples:

- Movie recommendations
- Product suggestions

---

## Anomaly Detection

Finding unusual patterns.

Examples:

- Detecting unusual financial transactions
- Finding network problems

---


# Semi-Supervised Learning (The Hybrid Approach)

What if labeling data is too expensive, but you have millions of unlabeled examples?

**Semi-Supervised Learning** combines a small amount of labeled data with a large amount of unlabeled data during training.

- **Photo Tagging (e.g., Apple/Google Photos):** The system automatically clusters similar-looking faces together (**Unsupervised**), and then asks you to provide a name label just once (**Supervised**) to tag that person across all photos.
- **Email Spam Detection:**
    -  Only a small number of emails are manually labeled as "spam" or "not spam."
    -  Millions of unlabeled emails help the model learn common patterns and improve spam detection.
- **Medical Image Analysis:**
    - Doctors label only a limited number of X-rays, MRIs, or CT scans because labeling is expensive.
    - The model uses thousands of unlabeled medical images to improve disease detection.
- **Speech Recognition:**
    - A small set of audio recordings has accurate transcripts.
    - A much larger collection of unlabeled speech recordings helps the model recognize accents pronunciations, and speech patterns.
- 

# 3. Reinforcement Learning

## Definition

Reinforcement Learning is a type of Machine Learning where an **agent learns by interacting with an environment**.

The agent makes decisions and receives feedback in the form of:

- Rewards
- Penalties

The goal is to learn actions that maximize long-term rewards.

---

# Reinforcement Learning Components

A reinforcement learning system contains:

## Agent

The learner or decision maker.

Example:

- Robot
- Game player
- Autonomous vehicle

---

## Environment

The world where the agent operates.

Examples:

- A game
- A road
- A simulated environment

---

## Action

A decision made by the agent.

Examples:

- Move left
- Accelerate
- Select an option

---

## Reward

Feedback received after an action.

Examples:

- Positive reward: Good decision
- Negative reward: Bad decision

---

# Reinforcement Learning Process

    Environment

         ↑
         
      Action

         ↑
         
      Agent

         ↓

      Reward


The agent learns through repeated interactions.

---

# Examples of Reinforcement Learning

## Game Playing

An AI agent learns to play a game.

Examples:

- Winning moves receive rewards.
- Losing moves receive penalties.

The system improves through experience.

---

## Robotics

A robot learns:

- How to move
- How to avoid obstacles
- How to complete tasks

---

## Autonomous Driving

A vehicle learns decision-making strategies:

- Speed control
- Lane selection
- Navigation decisions

---

# Comparing Learning Types

| Feature | Supervised Learning | Unsupervised Learning | Reinforcement Learning |
|---------|---------------------|-----------------------|------------------------|
| Data Type | Labeled data | Unlabeled data | Interaction data |
| Learning Method | Learn from examples | Find hidden patterns | Learn from rewards |
| Human Feedback | Direct labels | No labels | Reward signals |
| Main Goal | Predict outputs | Discover structures | Maximize rewards |
| Output | Prediction | Groups or patterns | Best actions |
| Example | Spam detection | Customer segmentation | Game playing |

---

# Choosing the Right Learning Approach

## Problem: Predict house price

Available:

- House features
- Historical prices

Solution:

**Supervised Learning**

Because we have labeled examples.

---

## Problem: Group customers by behavior

Available:

- Customer activity data
- No predefined groups

Solution:

**Unsupervised Learning**

Because we want to discover patterns.

---

## Problem: Train a robot to walk

Available:

- Environment
- Rewards for successful actions

Solution:

**Reinforcement Learning**

Because the agent learns through interaction.

---

# Key Takeaways

- Supervised Learning learns from labeled examples.
- Unsupervised Learning discovers hidden patterns from unlabeled data.
- Reinforcement Learning learns through interaction and rewards.
- The type of data and the goal of the problem determine the appropriate learning approach.

In the next section, we will learn what a **Machine Learning model** is and how a model learns from data.

# Model Evaluation, Underfitting, and Overfitting

## Learning Objectives

By the end of this section, you should be able to:

- Understand why a Machine Learning model must be evaluated
- Distinguish between training and testing performance
- Define underfitting and overfitting
- Explain why both are undesirable
- Understand the concept of model generalization

---

# Why Do We Evaluate a Model?

Training a Machine Learning model is only the first step.

The real question is:

> **How do we know if the model has learned correctly?**

A model should not only perform well on the data it has already seen, but also make accurate predictions on **new, unseen data**.

This ability is called **generalization**.

---

# Generalization

A good Machine Learning model learns the **underlying patterns** in the data rather than simply memorizing the training examples.

```
Training Data
      ↓
Learning
      ↓
Good Model
      ↓
New Data
      ↓
Accurate Predictions
```

The primary goal of Machine Learning is **good generalization**.

---

# Training Performance vs Testing Performance

When evaluating a model, we usually compare its performance on two datasets.

## Training Data

The data used to teach the model.

The model has already seen these examples during learning.

---

## Test Data

New data that was **never shown** during training.

The model must make predictions using only what it has learned.

---

A good model should perform well on **both** datasets.

---

# Measuring Model Performance

To evaluate a model, we compare:

- The model's prediction
- The correct answer (Ground Truth)

Example:

| Actual | Prediction |
|---------|------------|
| Spam | Spam |
| Not Spam | Spam |
| Spam | Spam |
| Not Spam | Not Spam |

Some predictions are correct.

Some are incorrect.

The more correct predictions, the better the model.

---

# Accuracy

One of the simplest evaluation metrics is **Accuracy**.

Accuracy measures the percentage of correct predictions.

Example:

```
100 predictions

92 correct

8 incorrect

Accuracy = 92%
```

Higher accuracy generally indicates better performance.

However, accuracy alone is **not always sufficient**, especially for imbalanced datasets.

In future lessons, we will learn additional evaluation metrics such as Precision, Recall, and F1 Score.

---

# Prediction Error

Instead of measuring correct predictions, we can also measure **prediction error**.

The objective during training is to minimize this error.

```
Smaller Error
        ↓
Better Predictions
```

Machine Learning algorithms continuously adjust their parameters to reduce prediction error.

---

# A Good Model

A good model should:

- Learn useful patterns
- Ignore random noise
- Perform well on unseen data

The challenge is finding the right balance.

---

# Underfitting

## Definition

**Underfitting** occurs when a model is **too simple** to learn the underlying patterns in the data.

It fails to capture important relationships.

As a result:

- Poor performance on training data
- Poor performance on test data

---

# Characteristics of Underfitting

- Model is too simple
- High prediction error
- Poor accuracy
- Cannot capture the complexity of the problem

```
Simple Model
      ↓
Learns Too Little
      ↓
Poor Predictions
```

---

# Example of Underfitting

Suppose house prices depend on:

- Area
- Location
- Number of bedrooms
- Age
- Neighborhood quality

But the model uses only:

```
House Price = Area
```

The model ignores important information.

Its predictions will often be inaccurate.

---

# Causes of Underfitting

Possible reasons include:

- Model is too simple
- Too few features
- Insufficient training
- Poor feature engineering

---

# How to Reduce Underfitting

Possible solutions:

- Use a more powerful model
- Train longer
- Add informative features
- Improve data representation

---

# Overfitting

## Definition

**Overfitting** occurs when a model learns the training data **too well**.

Instead of learning general patterns, it memorizes the training examples, including random noise.

As a result:

- Excellent performance on training data
- Poor performance on unseen data

---

# Characteristics of Overfitting

- Model is too complex
- Very low training error
- High testing error
- Poor generalization

```
Complex Model
       ↓
Memorizes Training Data
       ↓
Fails on New Data
```

---

# Example of Overfitting

Imagine a teacher gives students exactly one set of practice questions.

Student A memorizes every answer without understanding the concepts.

During the practice exam, the student gets every question correct.

However, in the real exam with different questions, the student performs poorly.

The student memorized instead of learning.

An overfitted model behaves in the same way.

---

# Causes of Overfitting

Common causes include:

- Model is too complex
- Too many parameters
- Too little training data
- Learning noise instead of patterns

---

# How to Reduce Overfitting

Common techniques include:

- Collect more training data
- Simplify the model
- Remove unnecessary features
- Stop training earlier (Early Stopping)
- Apply Regularization
- Use Data Augmentation (for images)
- Validate the model during training

These techniques will be discussed later in the course.

---

# Underfitting vs Overfitting

| Underfitting | Good Fit | Overfitting |
|---------------|----------|-------------|
| Too simple | Balanced complexity | Too complex |
| Learns too little | Learns useful patterns | Memorizes training data |
| High training error | Low training error | Very low training error |
| High testing error | Low testing error | High testing error |
| Poor generalization | Good generalization | Poor generalization |

---

# Visual Interpretation

```
                 Model Complexity
                        →

Underfitting ---- Good Fit ---- Overfitting
```

As model complexity increases:

- Performance first improves.
- Eventually it becomes too complex.
- Generalization begins to decrease.

---

# Generalization is the Goal

The objective of Machine Learning is **not** to memorize the training dataset.

The objective is to build models that perform well on **new, unseen data**.

A successful model:

- Learns meaningful patterns
- Ignores random noise
- Makes reliable predictions on future data

---

# Summary

| Concept | Description |
|----------|-------------|
| Evaluation | Measuring how well a model performs |
| Training Data | Data used for learning |
| Test Data | Unseen data used for evaluation |
| Accuracy | Percentage of correct predictions |
| Error | Difference between prediction and ground truth |
| Underfitting | Model is too simple to learn the data |
| Overfitting | Model memorizes the training data |
| Generalization | Ability to perform well on unseen data |

---

# Key Takeaways

- Training performance alone is **not enough**.
- A model should perform well on unseen data.
- Underfitting happens when the model learns too little.
- Overfitting happens when the model memorizes the training data.
- The goal is to achieve **good generalization**, where the model captures the true patterns in the data without memorizing noise.

---

# Example 1: House Price Prediction

Suppose we want to predict house prices using the following features:

- Area
- Number of bedrooms
- Age of the house

We train three different models.

## Model A

Predictions:

| Actual Price | Predicted Price |
|--------------|----------------|
| 250,000 | 260,000 |
| 320,000 | 300,000 |
| 410,000 | 395,000 |

The model makes reasonable predictions for both training and new houses.

✅ This is a **Good Fit**.

---

## Model B

Predictions:

| Training Data | Error |
|--------------|-------|
| House 1 | High |
| House 2 | High |
| House 3 | High |

The model cannot even fit the training data.

Its predictions are poor for every house.

❌ This is **Underfitting**.

Reason:

- The model is too simple.

---

## Model C

Training Results:

| House | Actual | Predicted |
|--------|--------|-----------|
| 1 | 250,000 | 250,000 |
| 2 | 320,000 | 320,000 |
| 3 | 410,000 | 410,000 |

Perfect!

Now test it on new houses:

| Actual | Predicted |
|---------|-----------|
| 350,000 | 450,000 |
| 280,000 | 180,000 |

Very poor performance.

❌ This is **Overfitting**.

The model memorized the training houses instead of learning general pricing patterns.

---

# Example 2: Student Exam Prediction

Suppose we want to predict whether a student will pass an exam.

Features:

- Study hours
- Attendance
- Previous grades

Label:

- Pass
- Fail

### Underfitting

The model predicts:

```
Every student will PASS.
```

Obviously, this is too simple.

Many predictions will be incorrect.

---

### Overfitting

The model memorizes every student from last year's class.

It predicts perfectly for those students.

However, when new students arrive next semester, its predictions become inaccurate.

The model learned the students instead of learning the relationship between study habits and exam success.

---

# Example 3: Spam Detection

Suppose we build a spam classifier.

Training emails:

```
Win a FREE iPhone!

Claim your prize!

Limited-time offer!
```

---

## Underfitting

The model simply labels every email as:

```
Not Spam
```

Many spam emails are missed.

---

## Overfitting

The model memorizes exact training emails.

When it receives:

```
Congratulations!
You have won a free laptop!
```

it predicts:

```
Not Spam
```

because this exact sentence was never seen before.

The model memorized instead of learning general spam patterns.

---

# Example 4: Face Recognition

Suppose we train a model using only ten pictures of one person.

### Underfitting

The model cannot even recognize those ten pictures correctly.

It has learned too little.

---

### Overfitting

The model recognizes only the exact training photos.

If the person changes:

- hairstyle
- glasses
- lighting
- camera angle

the model fails.

It memorized the training images instead of learning facial features.

---

# Analogy: Studying for an Exam

Imagine three students preparing for the final exam.

## Student A — Underfitting

- Reads only one page.
- Doesn't understand the concepts.
- Performs poorly on the exam.

This is **Underfitting**.

---

## Student B — Good Fit

- Understands the concepts.
- Solves many different exercises.
- Can answer new questions.

This is **Good Generalization**.

---

## Student C — Overfitting

- Memorizes every homework solution.
- Gets perfect scores on homework.
- Cannot solve new exam questions.

This is **Overfitting**.

---

# Comparing the Three Cases

| Situation | Training Performance | Test Performance | Conclusion |
|-----------|----------------------|------------------|------------|
| Underfitting | Poor | Poor | Model is too simple |
| Good Fit | Good | Good | Desired model |
| Overfitting | Excellent | Poor | Model memorizes data |

---

# Discussion

Consider the following applications.

For each one, discuss what underfitting and overfitting might look like.

- House price prediction
- Email spam detection
- Disease diagnosis
- Face recognition
- Movie recommendation

**Questions**

1. What would an underfitted model do?
2. What would an overfitted model do?
3. Which model would you deploy in a real application?
