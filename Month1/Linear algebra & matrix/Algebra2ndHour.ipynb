## 1. What is a Vector?

In Machine Learning, a **vector** is just a list of numbers representing the **features** of a single item or sample.

For example, a user's ratings for 3 movies can be a vector:
- User A rated: Action(5), Comedy(2), Drama(4)
- Vector A = $[5, 2, 4]$

In NumPy, a vector is simply a 1D array.

```python
import numpy as np

user_A = np.array([5, 2, 4])
user_B = np.array([4, 3, 5])

print("User A:", user_A)
print("User B:", user_B)
```

## 2. Vector Arithmetic

Because they are NumPy arrays, we can easily perform mathematical operations element-by-element.

- **Addition & Subtraction:** Done element-wise.
- **Scalar Multiplication:** Multiplying the whole vector by a single number.

```python
# Addition
print("A + B =", user_A + user_B)

# Subtraction
print("A - B =", user_A - user_B)

# Scalar Multiplication
print("A * 2 =", user_A * 2)
```

## 3. Euclidean Distance: Measuring Similarity

How do we know if two users have similar tastes? We calculate the distance between their vectors in space.

The most common way is the **Euclidean Distance**. The formula is:

$$ \mathbf{A} = [ A_1, A_2, \cdots, A_n ]$$
$$ \mathbf{B} = [ B_1, B_2, \cdots, B_n ]$$

$$d = \sqrt{\sum_{i=1}^{n} (A_i - B_i)^2}$$


Similarit is almost the inverse of $$s \approx \frac{1}{d}$$


In plain English:
1. Subtract vector B from vector A
2. Square the differences
3. Sum all the squared differences
4. Take the square root of the sum

Let's implement this step-by-step in NumPy!

```python
import math

def find_similarity(v1, v2):
    # Step 1: Subtract
    diff = v1 - v2
    print("Difference:", diff)
    
    # Step 2: Square the differences
    squared_diff = diff ** 2
    print("Squared differences:", squared_diff)
    
    # Step 3: Sum them up
    sum_squared = np.sum(squared_diff)
    print("Sum of squares:", sum_squared)
    
    # Step 4: Square root
    distance = np.sqrt(sum_squared)
    print("Euclidean Distance:", distance)
    
    
    # Step 5: Invert the distance
    similarity = 1.0 / (distance + 1e-8)

    return similarity

s = find_similarity(user_A, user_B)
print(f"Similarity between user A and user B:{s}")
```

## 4. The NumPy Shortcut (`np.linalg.norm`)

Instead of writing those 4 lines every time, NumPy has a built-in function to calculate this distance (also known as the vector norm).

```python
# Calculating distance in one line!
quick_distance = np.linalg.norm(user_A - user_B)
print("Distance using linalg.norm:", quick_distance)
```

## 4. The Cosine Similarity

Cosine Similarity measures how similar two vectors are by comparing the angle between them, rather than their lengths. 

More explanations in the near future. Don't worry :)

```python
dot_product = np.dot(user_A, user_B)
norm_A = np.linalg.norm(user_A)
norm_B = np.linalg.norm(user_B)

cosine_similarity = dot_product / (norm_A * norm_B)

print("Cosine Similarity:", cosine_similarity)
```

## 5. Mini-Project: A Simple Recommendation System

Let's say we have a target user, and we want to find out which other user is most similar to them so we can recommend movies.

**Rule:** The *smaller* the distance, the *more similar* they are.

```python
# Ratings for [Action, Comedy, Drama, Sci-Fi]
target_user = np.array([5, 1, 1, 5])

# Database of other users
user_1 = np.array([4, 2, 1, 4])  # Loves Action and Sci-Fi
user_2 = np.array([1, 5, 5, 2])  # Loves Comedy and Drama
user_3 = np.array([5, 1, 2, 4])  # Loves Action and Sci-Fi

# Calculate distances
dist_1 = np.linalg.norm(target_user - user_1)
dist_2 = np.linalg.norm(target_user - user_2)
dist_3 = np.linalg.norm(target_user - user_3)

print(f"Distance to User 1: {dist_1:.2f}")
print(f"Distance to User 2: {dist_2:.2f}")
print(f"Distance to User 3: {dist_3:.2f}")
```

**Conclusion:** The distance to User 3 is the smallest. Therefore, `target_user` is most similar to `user_3`. If `user_3` likes a new Action movie, we should recommend it to `target_user`!
