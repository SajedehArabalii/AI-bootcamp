## 4. Initialization: Common Ways to Create NumPy Arrays
While you can turn Python lists into arrays using `np.array()`, Machine Learning algorithms frequently require initializing arrays from scratch (e.g., creating blank matrices for weights, bias vectors set to zero, or tracking values over a linear range). 

Here are the 6 structural array creation tools you must know:
1. `np.array()`: Converts existing Python structures (like lists) into arrays.
2. `np.zeros(shape)`: Fills an array of a given shape with `0.0`. (Commonly used to initialize neural network *biases*).
3. `np.ones(shape)`: Fills an array of a given shape with `1.0`. 
4. `np.eye(n)`: Creates a 2D **Identity Matrix** of size $n \times n$ (ones on the main diagonal, zeros everywhere else).
5. `np.arange(start, stop, step)`: Creates a sequence of values with a defined *step size* (similar to Python's `range()`).
6. `np.linspace(start, stop, num)`: Creates a sequence with a specified *number of elements*, automatically calculating equal spacing.

```python
import numpy as np

# 1. Custom Array from List
custom_arr = np.array([5, 10, 15])
print("1. np.array:\n", custom_arr)
```

```python
# 2. Array of Zeros (Often used for Model Biases)
bias_vector = np.zeros(shape=(3,3,3)) 
print("\n2. np.zeros (1D vector):\n", bias_vector)
```

```python
# 3. Matrix of Ones
ones_matrix = np.ones(shape=(2, 4))
print("\n3. np.ones (2x4 matrix):\n", ones_matrix)
```

```python
# 4. Identity Matrix
identity_matrix = np.eye(3)
print("\n4. np.eye (3x3 Identity):\n", identity_matrix)
```

```python
# 5. Arrange by Step Size
step_sequence = np.arange(start=0, stop=20, step=1)
print("\n5. np.arange [0 to 10, step 2]:\n", step_sequence)
```

```python
# 6. Linspace by Number of Elements (Great for creating plotting intervals)
linear_intervals = np.linspace(start=0, stop=1, num=4)
print("\n6. np.linspace [0 to 1 split into 5 pieces]:\n", linear_intervals)
```

## Computing  $sin(x)e^{-x}$

[sin(x[i]) * exp(-x[i]) for i in range(n)]

np.sin(x) * np.exp(-x)

```python
import math

n = 100000
x = np.linspace(0, 2*np.pi, n)

# list comprehension
%timeit y = [math.sin(x[i]) * math.exp(-x[i]) for i in range(n)]
# use numpy ufuncs
%timeit y = np.sin(x) * np.exp(-x)
```

**What is Vectorization?**

Vectorization refers to one or both of the following, depending on the context, both of which are used in NumPy’s universal functions (ufuncs):

(1) Applying operations to all elements of a collection at once.

(2) Utilizing hardware instruction sets for parallel computation of a single instruction (SIMD).

```python
def f(v): # apply f to one scalar value v
    return math.sin(v) * math.exp(-v)

vectorizedf = map(f, x) # معادل قابلیت اول موجود در نامپای

F = np.vectorize(f)

# list comprehension
%timeit y = [f(x[i]) for i in range(n)]

# use map
# %timeit y = list(vectorizedf)
%timeit y = list(map(f, x))

# use numpy's vectorize
%timeit y = F(x)

# use numpy's ufunc
%timeit y = np.sin(x) * np.exp(-x)
```

## 12. Indexing and Slicing in 2D Arrays

```python
dataset = np.array([
    [2000, 3, 10, 500000],
    [1500, 2, 5, 350000],
    [1800, 3, 8, 420000],
    [1900, 4, 10, 460000],
    [2200, 4, 12, 520000]
])

print(dataset)
print(dataset.shape)
```

```python
print("First row:", dataset[0, :])
print("last row:", dataset[-1, :])
print("even rows:", dataset[0:5:2, :])
print("odd rows:", dataset[1:7:2, :])

print("All rows, first column:", dataset[:, 0])
print("First sample, second feature:", dataset[0, 1])
print("All rows, last column:", dataset[:, -1])
```

## 13. Splitting a Dataset into Features and Target

This is one of the most important uses of slicing in Machine Learning.

If the last column is the target:
- `X` = all rows, all columns except the last one
- `y` = all rows, only the last column

```python
X = dataset[:, :-1]
y = dataset[:, -1]

print("X:\n", X)
print("\ny:\n", y)
```

Now we have:
- `X`: the input features
- `y`: the target values we want to predict

This simple slicing idea appears everywhere in data science and machine learning.

```python
print("X shape:", X.shape)
print("y shape:", y.shape)
```

## 15. Altering Array Shapes & Random Initialization
In Machine Learning pipelines, raw data often doesn't match the dimensions your models or neural networks expect. We can alter shapes using `.reshape()` or `.T` (Transpose) to flip rows and columns. We also often initialize parameters using random distributions.

- `.reshape(rows, cols)`: Changes dimensions without changing the underlying data.
- `.T`: Transposes a 2D matrix (swaps rows and columns).
- `np.random.randn(rows, cols)`: Generates an array filled with random numbers from a standard normal distribution (mean=0, std=1)—essential for weight initializations.
- `np.random.seed(42)`: Fixes the random numbers generated so your code yields identical results every run.

```python
# Fix randomness for reproducible results
np.random.seed(42)

# Generate a flat vector of 6 elements
flat_data = np.array([10, 20, 30, 40, 50, 60])
print("Original Vector:", flat_data)

# Reshape into a 3x2 Matrix (3 rows, 2 features)
matrix_data = flat_data.reshape(3, 2)
print("\nReshaped into 3x2 Matrix:\n", matrix_data)

# Transpose the matrix into a 2x3 Matrix
print("\nTransposed Matrix (.T):\n", matrix_data.T)

# Generate a completely random weight matrix for a neural network
random_weights = np.random.randn(2, 3)
print("\nRandom Weight Matrix (2x3):\n", random_weights)
```
