# Hour 4: Dot Product and the Birth of a Linear Neuron

In this section, we will get acquainted with one of the most important mathematical operations in AI, namely the **Dot Product**, and build our first simple model (a linear neuron).

```python
import numpy as np
```

## 1. Element-wise Multiplication vs. Dot Product

So far, we have seen that the `*` operator performs element-wise multiplication on arrays. However, in neural networks, we are looking to calculate a **weighted sum**. For this, we use the dot product with the `np.dot` function or the `@` operator.

```python
x = np.array([2, 4, 6]) # Features
w = np.array([0.5, 1.5, 0.1]) # Weights

print("Element-wise multiplication (*):", x * w)
print("Dot product (@):", x @ w)  # (2*0.5) + (4*1.5) + (6*0.1)
```

## 2. Linear Neuron: Adding Bias

The basic formula of a neuron is as follows: $$ y = x \cdot w + b $$
The bias ($b$) acts like a starting point or y-intercept.

<img src="linear.png" width="400">

<img src="nonlinear.png" width="400">

```python
x = np.array([2, 4, 6]) # Features
w = np.array([0.5, 1.5, 0.1]) # Weights
b = 2.0 # Bias
z = (x @ w) + b
print("Neuron output for a single sample:", z)
```

## 3. Batch Processing

Instead of calculating one by one, we put the features of multiple samples into a matrix $X$ and calculate the output for all of them in a single line of code! Here, Broadcasting comes to our aid to add the bias to everything.

```python
w = np.array([0.5, 1.5, 0.1]) # Features

x1 = np.array([2, 4, 6])
x2 = np.array([1, 5, 2])
x3 = np.array([3, 1, 9])

print(f"similarity of w to x1: {x1 @ w}")
print(f"similarity of w to x2: {x2 @ w}")
print(f"similarity of w to x3: {x3 @ w}")
```

```python
# 3 samples, 3 features each
X = np.array([
    [2, 4, 6],
    [1, 5, 2],
    [3, 1, 9]
])
b = 0
Z = (X @ w) + b
print("Neuron output for a Batch of data:\n", Z)
```

# Why the Dot Product Measures Similarity

<img src="dot.png" width="400">

The geometric definition of the dot product is


$$\mathbf{a} \cdot \mathbf{b} = |\mathbf{a}|\,|\mathbf{b}| \cos\theta$$

where:

- $|\mathbf{a}|$ is the magnitude (length) of **a**.
- $|\mathbf{b}|$ is the magnitude (length) of **b**.
- $\theta$ is the angle between the two vectors.

The key idea is the **$\cos\theta$** term, which depends only on the angle between the vectors.

---

- if $\theta=0  \rightarrow \mathbf{a}.\mathbf{b} = |\mathbf{a}|\,|\mathbf{b}|   $

- if $\theta < 90   \rightarrow 0 < \cos \theta < 1 \rightarrow \mathbf{a}.\mathbf{b} < |\mathbf{a}|\,|\mathbf{b}|   $

- if $\theta = 90   \rightarrow \cos \theta = 0 \rightarrow \mathbf{a}.\mathbf{b}  = 0   $

$ cs = \frac{\mathbf{a}.\mathbf{b}}{|\mathbf{a}|\,|\mathbf{b}|}  $

$ cs = (\frac{\mathbf{a}}{|\mathbf{a}|}).(\frac{\mathbf{b}}{|\mathbf{b}|})  $



```python
s1 = np.array([2 , 0])
s2 = np.array([0 , 2])
s3 = np.array([2 , 2])
s4 = np.array([-2 , 0])
```

```python
print(f"s1@s2:{s1@s2}")
```

```python
print(f"s1@s3:{s1@s3}")
```

```python
print(f"s1@s4:{s1@s4}")
```

```python
print(f"s1@s1:{s1@s1}")
```

```python
cs_1_3 = s1@s3/(np.linalg.norm(s1)*np.linalg.norm(s3))
print(f"cs_1_3:{cs_1_3}")
```

```python
cs_1_4 = s1@s4/(np.linalg.norm(s1)*np.linalg.norm(s4))
print(f"cs_1_4:{cs_1_4}")
```

```python
cs_1_1 = s1@s1/(np.linalg.norm(s1)*np.linalg.norm(s1))
print(f"cs_1_1:{cs_1_1}")
```

```python
# np.dot and @ are the same

cs_1_1 = np.dot(s1, s1)/(np.linalg.norm(s1)*np.linalg.norm(s1))
print(f"cs_1_1 with np.dot:{cs_1_1}")
```
