## 1. Shape Operations: Reshape and Transpose

In Machine Learning, your data's shape often doesn't perfectly match the shape your neural network expects. 

*   **Reshape (`np.reshape` or `.reshape()`)**: Changes the dimensions of an array without changing its data. A common use case is "flattening" a 2D image into a 1D vector.
*   **Transpose (`.T`)**: Swaps rows and columns. This is often used to align matrix inner dimensions for matrix multiplication.

````python
# Example: Flattening a 2x2 image into a 1D array of 4 pixels
image = np.array([
    [0.1, 0.9, 0.1],
    [0.8, 0.2, 0.2]
]) # Shape: (2, 3)

flattened_image = image.reshape(6) # -1 means "figure out the exact dimension automatically"
print("Original Image Shape:", image.shape)
print("Flattened Image:", flattened_image)
print("Flattened Shape:", flattened_image.shape)

print("-" * 30)

# Example: Transposing a Matrix
A = np.array([[1, 2, 3], [4, 5, 6]]) # Shape: (2, 3)
A_T = A.T                            # Shape: (3, 2)
print("Original A:\n", A)
print("Transposed A:\n", A_T)
````

````python
A = np.arange(24).reshape(2, 3, -1)
print(A)
print(A.shape)
# (2, 3, 4)
print(A.T.shape)
# (4, 3, 2)
````

## 2. Vectorized If-Else with `np.where`

Using Python `for` loops and `if` statements is too slow for large datasets. NumPy provides `np.where(condition, value_if_true, value_if_false)` to apply conditions across an entire array instantly.

````python
data = np.array([-2, 5, -8, 10, -1])

# Replace all negative numbers with 0, and keep positive numbers as they are
cleaned_data = np.where(data < 0, 0, data) # cleaned_data = max(0, data)

print("Original Data:", data)
print("Cleaned Data: ", cleaned_data)
````

## 3. ML Connection: The ReLU Activation Function

In neural networks, linear operations ($X @ W + b$) alone can only solve linear problems. To learn complex patterns, we pass the output through an **Activation Function**.

The most popular activation function is **ReLU (Rectified Linear Unit)**. 
Formula: $$ ReLU(x) = \max(0, x) $$
In simple terms: If the value is negative, make it $0$. If it's positive, leave it alone. This is perfectly handled by `np.where`!

````python
Z = np.array([[-1.5, 2.0, -0.5],
              [3.1, -2.2, 0.8]])

# Apply ReLU Activation
# A = np.where(Z > 0, Z, 0)
A = np.where(Z < 0, 0, Z)

print("Raw Output (Z):\n", Z)
print("\nActivated Output (A):\n", A)
````

## 4. Multi-Class Classification: Softmax & Structural Dimension Safety

While ReLU is great for hidden layers, multi-class classification networks require outputs to behave like probabilities (values between 0 and 1 that sum up to 1). For this, we use the **Softmax** activation function.

Formula for a vector element:
$$ Softmax(z_i) = \frac{e^{z_i}}{\sum e^{z_j}} $$

To calculate this across an entire batch matrix, we use exponential universal functions (`np.exp()`) combined with strict row-wise operations.

### The Crucial Role of `keepdims=True`
When summing probabilities across rows, omitting `keepdims=True` drops the 2D matrix structure, converting it into a 1D array. This causes subsequent matrix broadcasting to fail or apply incorrectly down columns instead of across rows.

````python
# Raw logits from a layer output: 2 samples, 3 classes
Z_logits = np.array([
    [2.0, 1.0, 0.1],
    [1.0, 3.0, 0.2]
])

# Step 1: Compute exponentials
exp_Z = np.exp(Z_logits)

# Step 2: Sum along rows (axis=1) keeping original 2D structural shape
sum_exp_Z = np.sum(exp_Z, axis=1, keepdims=True)

# Step 3: Use broadcasting to divide safely
probabilities = exp_Z / sum_exp_Z

print("Raw Matrix Shape:", Z_logits.shape)
print("Row Sum Matrix Shape (with keepdims=True):", sum_exp_Z.shape)
print("\nOutput Class Probabilities (Rows sum perfectly to 1.0):\n", probabilities)
````

## 5. Interpreting Predictions (`np.argmax`) & Memory Safety

### A. Turning Probabilities into Class Predictions
Once your network outputs final probabilities, use `np.argmax(axis=1)` to extract the index of the highest value in each row, yielding the final model classification.

````python
# Simulated model probabilities for 3 samples across 3 classes
# تشخیص نوع وسیله نقلیه از روی تصویر
# خروجی شبکه یک سه تایی (سبک، سنگین، موتور سیکلت)
model_outputs = np.array([
    [0.1, 0.7, 0.2],  # Sample 1 -> Predicted Class Index 1
    [0.8, 0.05, 0.15], # Sample 2 -> Predicted Class Index 0
    [0.3, 0.2, 0.5]   # Sample 3 -> Predicted Class Index 2
])

max_scores = np.max(model_outputs, axis=1)
print("Max Scores for the Batch:", max_scores)

# Extract the index of the max value across the columns (axis=1)
predicted_classes = np.argmax(model_outputs, axis=1)
print("Predicted Class Labels for the Batch:", predicted_classes)
````

### B. Array Views vs. Copies
In NumPy, slicing or assigning an array variable does not create a new copy of the data. Instead, it creates a "View." Modifying a slice or batch assignment will inadvertently alter your original underlying dataset!

Use ```.copy()``` when you explicitly need to decouple intermediate variables or mini-batches from your primary dataset.

````python
# Modifying a View (Dangerous Trap)
original_dataset = np.array([10, 20, 30, 40])
print("Original Dataset:", original_dataset)
view_slice = original_dataset[0:2]
view_slice[0] = 999
print("Original Dataset altered by view modification:", original_dataset)
````

````python
# Modifying an Explicit Copy (Safe)
original_dataset = np.array([10, 20, 30, 40])
print("Original Dataset:", original_dataset)

safe_copy = original_dataset[0:2].copy()
safe_copy[0] = 999
print("Original Dataset preserved when using .copy():", original_dataset)
````
